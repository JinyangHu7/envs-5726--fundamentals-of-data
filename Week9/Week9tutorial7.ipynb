```python
from pathlib import Path
import csv
import matplotlib.pyplot as plt
```

Second step: Load the data

```python
csv_path = Path(
    "/Users/admin/Downloads/EMDAT_Haiti_Disaster_Types_Summary.csv")

emdat_summary_table = []

with open(csv_path, "r", encoding="utf-8") as csv_file:
    reader = csv.reader(csv_file)
    headers = next(reader)

    for row in reader:
        emdat_summary_table.append(row)

category_column = [
    row[headers.index("Disaster Type")]
    for row in emdat_summary_table
]

value_column = [
    int(row[headers.index("Total Deaths")])
    for row in emdat_summary_table
]
```

Last step: Plot the pie chart

```python
plt.pie(value_column, labels = category_column)

plt.title("Total Deaths by Natural Disaster Type")

plt.legend(bbox_to_anchor=(1.2, 0.6), loc="upper left")

plt.show()
```
